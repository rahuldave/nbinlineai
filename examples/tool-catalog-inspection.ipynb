# Tool catalog: live Python

Run setup to create a small score list, a log, a doubling function, and a disposable registered skill. The sections below inspect those **live kernel values** and installed standard-library objects; they do not read a saved notebook. Compare a signature, a bounded value, a source location, and a short trace.


```python
from nbinlineai.tools import (
    search_kernel_names, inspect_python, show_doc, api_names, search_docs, inspect_value, search_value, source_files, list_skills, read_skill, trace_function
)

catalog_scores = [2, 4, 6]
catalog_log = "ready\nno error\nready"
def catalog_double(value: int) -> int:
    """Double a small integer for the trace example."""
    return value * 2
# A small, registered pyskill entry point lets both discovery and static reading
# show a real installed-style module without inspecting a personal skill.
import sys
from pathlib import Path
from tempfile import TemporaryDirectory
catalog_skill_temp = TemporaryDirectory(prefix="nbinlineai-skill-")
catalog_skill_root = Path(catalog_skill_temp.name)
(catalog_skill_root / "nbinlineai_demo_skill.py").write_text(
    "\"\"\"Read the score list first; never change the learner's scores.\"\"\"\n"
    "def main():\n    return 'Study the three scores.'\n", encoding="utf-8")
catalog_skill_dist = catalog_skill_root / "nbinlineai_demo_skill-1.0.dist-info"
catalog_skill_dist.mkdir()
(catalog_skill_dist / "METADATA").write_text(
    "Metadata-Version: 2.1\nName: nbinlineai-demo-skill\nVersion: 1.0\n", encoding="utf-8")
(catalog_skill_dist / "entry_points.txt").write_text(
    "[pyskills]\nnbinlineai_demo = nbinlineai_demo_skill:main\n", encoding="utf-8")
sys.path.insert(0, str(catalog_skill_root))
print("Disposable registered skill: nbinlineai_demo")
```

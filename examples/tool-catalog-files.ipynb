# Tool catalog: saved files and source

Run setup to create a temporary Python project and notes. Follow the sections in order: inspect existing files, search the source, read an outline, then make checked edits using paths, section addresses, and digests returned **in this run**. Every write stays in the temporary directory. Its files disappear when this kernel exits.


```python
from pathlib import Path
from tempfile import TemporaryDirectory
import hashlib
from nbinlineai.tools import tool_catalog
from nbinlineai.tools import (
    path_info, list_files, view_file, create_file, file_str_replace, file_insert_line, file_replace_lines, search_files, source_doc, document_outline, read_document_section, file_strs_replace, view_file_hashes, file_replace_checked
)

catalog_temp = TemporaryDirectory(prefix="nbinlineai-catalog-")
catalog_root = Path(catalog_temp.name)
catalog_source = catalog_root / "sample.py"
catalog_source.write_text('"""Disposable source."""\n\ndef double(value: int) -> int:\n    """Double one value."""\n    return value * 2\n', encoding="utf-8")
catalog_notes = catalog_root / "notes.md"
catalog_notes.write_text("# Field notes\nOne observation.\n## Counts\nvisits = 2\n", encoding="utf-8")
catalog_edit = catalog_root / "edit.txt"
catalog_edit.write_text("alpha\nbeta\ngamma\n", encoding="utf-8")
print("Disposable folder:", catalog_root)
# Independent paths keep the AI edits separate from the direct-call examples.
catalog_ai_create_path = str(catalog_root / "ai-created.txt")
catalog_ai_replace_path = str(catalog_root / "ai-replace.txt")
catalog_ai_insert_path = str(catalog_root / "ai-insert.txt")
catalog_ai_lines_path = str(catalog_root / "ai-lines.txt")
catalog_ai_multi_path = str(catalog_root / "ai-multi.txt")
catalog_ai_checked_path = str(catalog_root / "ai-checked.txt")
for path in (catalog_ai_replace_path, catalog_ai_insert_path,
             catalog_ai_lines_path, catalog_ai_multi_path, catalog_ai_checked_path):
    Path(path).write_text("alpha\nbeta\ngamma\n", encoding="utf-8")
print("Five separate disposable files are ready for AI edits.")
```

```python
print(tool_catalog("files"))
```

## path_info


```python
print(path_info(str(catalog_root)))
```

## list_files


```python
print(list_files(str(catalog_root), pattern="*.py"))
```

## view_file


```python
print(view_file(str(catalog_source), 1, 5))
```

## create_file


```python
print(create_file(str(catalog_root / "created.txt"), "Created by the example.\n"))
```

## file_str_replace


```python
print(file_str_replace(str(catalog_edit), "alpha", "first", expected_matches=1))
```

## file_insert_line


```python
print(file_insert_line(str(catalog_edit), 1, "between"))
```

## file_replace_lines


```python
print(file_replace_lines(str(catalog_edit), 2, 2, "second"))
```

## search_files


```python
print(search_files("double", path=str(catalog_root), pattern="*.py"))
```

## source_doc


```python
print(source_doc(str(catalog_source), symbol="double"))
```

## document_outline


```python
catalog_outline = document_outline(str(catalog_notes))
print(catalog_outline)
```

## read_document_section


```python
catalog_counts_address = next(line.split()[0] for line in catalog_outline.splitlines() if "Counts (lines" in line)
print(read_document_section(str(catalog_notes), section=catalog_counts_address))
```

## file_strs_replace


```python
print(file_strs_replace(str(catalog_edit), ["first", "gamma"], ["one", "three"]))
```

## view_file_hashes


```python
catalog_hashes = view_file_hashes(str(catalog_edit), 1, 6)
print(catalog_hashes)
```

## file_replace_checked


```python
catalog_digest = catalog_hashes.splitlines()[1].split(": ", 1)[1]
print(file_replace_checked(str(catalog_edit), "three", "3", catalog_digest))
```

## Ask AI to use the same file tools

Run these questions after the direct code examples. They use separate temporary files, so every edit has a fresh, checkable starting state. Each question explicitly names its tools and reports the returned evidence.

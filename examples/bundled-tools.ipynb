# Bundled notebook tools

nbinlineai ships ten optional tools. This lesson focuses on four read-only kernel and saved-notebook tools: search live Python **names**, list saved notebooks, find saved notebook cells, and read a saved cell's source. The live-cell and web tools are introduced in the other example notebooks. Choose your own provider and model in **Configure AI**. Run the setup cells before the AI Prompts. No provider key or generated answer is saved here.

These tools are available to the AI **only when their `&` references appear in the AI Prompt being executed**. A list in an ordinary Markdown note or an earlier AI prompt does not register them for a later request. The `tools_markdown` helper prints references to copy; it is not itself a tool.

```python
from pathlib import Path

from nbinlineai.tools import (
    find_notebook_cells,
    insert_markdown,
    inspect_python,
    list_cells,
    list_notebooks,
    read_cell,
    read_notebook_cell,
    read_url,
    search_kernel_names,
    tools_markdown,
    url_to_note,
)

study_roster_marker: dict[str, int] = {"Asha": 2, "Mina": 3}
fixture_candidates: list[Path] = [
    Path("examples/data/ecosystem-lesson.ipynb"),
    Path("data/ecosystem-lesson.ipynb"),
]
fixture_path: str = ""
for candidate in fixture_candidates:
    if candidate.is_file():
        fixture_path = str(candidate)
        break
if not fixture_path:
    raise FileNotFoundError(
        "Copy the examples/data folder alongside this notebook, or run from the repository root."
    )
fixture_directory: str = str(Path(fixture_path).parent)
print(f"Saved fixture: {fixture_path}")
```

## Generate selected or all tool references

The selected list exposes only the two saved-notebook cell tools; the all-tools list prints all ten imported built-ins. The other six are explained in the live-notebook and Python/web lessons. The helper itself is **not** callable by AI.

## Live names are not live values

The next prompt registers only `search_kernel_names` and asks it to find a distinctive name in the running Python kernel. This tool lists matching names; use `$` syntax in a later prompt if you want a named variable's live value. It does not expose every kernel value automatically.

Call &`search_kernel_names` with query `study_roster_marker`. Report the matching Python name only; do not guess its value.

## List saved notebooks in the fixture directory

This prompt exposes only `list_notebooks`. The tool reads saved `.ipynb` paths under the directory printed by the setup cell; it does not open an unsaved notebook in the browser. It can discover the fixture that the next prompt searches.

Call &`list_notebooks` with `path` set to the live $`fixture_directory`. Report the saved notebook filename you find there.

## Inspect source in a saved notebook

The next prompt registers two tools and reads the checked-in fixture on disk. It first searches for “lichen indicator,” then reads that returned cell ID. A saved `.ipynb` path is required: the tools do **not** read unsaved edits in JupyterLab, rich code outputs, or image bytes. The live notebook cell above prints the path that `$` interpolation will provide.

The saved fixture path is $`fixture_path`. Call &`find_notebook_cells` on that path with query `lichen indicator`. Then call &`read_notebook_cell` with the same path and the matching cell ID. Summarize what the saved source actually says, and name the cell ID you read.

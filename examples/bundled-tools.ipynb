# Bundled notebook tools

nbinlineai ships ten optional tools. This lesson focuses on four read-only kernel and saved-notebook tools: search live Python **names**, list saved notebooks, find saved notebook cells, and read a saved cell's source. The live-cell and web tools are introduced in the other example notebooks. Choose your own provider and model in **Configure AI**. Run the setup cells before the AI questions. No provider key or generated answer is saved here.

The ordinary Markdown declaration below offers four selected tools to every lower AI question. The imported functions must also be present in the live kernel. A later question does not need to repeat an `&` reference. The `tools_markdown` helper prints references you can use to create or edit declarations; it is not itself a tool.


```python
from pathlib import Path

from nbinlineai.tools import (
    find_notebook_cells,
    insert_markdown,
    inspect_python,
    list_cells,
    list_notebooks,
    read_cell,
    read_notebook_cell,
    read_url,
    search_kernel_names,
    tools_markdown,
    url_to_note,
)

study_roster_marker: dict[str, int] = {"Asha": 2, "Mina": 3}
fixture_candidates: list[Path] = [
    Path("examples/data/ecosystem-lesson.ipynb"),
    Path("data/ecosystem-lesson.ipynb"),
]
fixture_path: str = ""
for candidate in fixture_candidates:
    if candidate.is_file():
        fixture_path = str(candidate)
        break
if not fixture_path:
    raise FileNotFoundError(
        "Copy the examples/data folder alongside this notebook, or run from the repository root."
    )
fixture_directory: str = str(Path(fixture_path).parent)
print(f"Saved fixture: {fixture_path}")
```

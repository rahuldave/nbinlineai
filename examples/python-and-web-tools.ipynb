# Inspect Python and read public documentation

The read-only Python inspector uses explicit live names, not arbitrary Python expressions. The web reader accepts public HTTP(S) pages only and includes its source URL. The note tool fetches a public page and inserts bounded Markdown into this notebook. Network requests occur only when you run the AI Prompts that offer those tools.

```
from nbinlineai.tools import inspect_python, read_url, tools_markdown, url_to_note


def calculate_average(values: list[int]) -> float:
    """Return the arithmetic mean of a nonempty list of integers."""
    return sum(values) / len(values)

average_alias = calculate_average
print("Python and web tools imported.")
```

```
selected = tools_markdown(
    ["inspect_python", "read_url", "url_to_note", "average_alias"],
    custom={"average_alias": average_alias},
)
print(selected)
```

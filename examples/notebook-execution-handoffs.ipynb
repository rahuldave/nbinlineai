# Queued notebook execution handoffs

This notebook uses the ongoing experimental Git branch. Run one section at a time in JupyterLab with nbinlineai loaded. Each handoff finishes its current AI question or Python cell before the next cell runs in the same kernel. Review new cells and save the notebook when you want to keep them. The AI questions need a configured provider; ordinary setup code does not.

```python
from nbinlineai.tools import add_code_cell_and_execute, prompt_and_run, run_and_prompt

scores = [2, 4, 6]
print(f"Ready with {len(scores)} scores")
```

## Offer the three handoff tools

- &`add_code_cell_and_execute` — select existing code by ID, or add and run new code.
- &`run_and_prompt` — run code, then ask a separate AI question about that run.
- &`prompt_and_run` — ask a separate AI question that must choose code before anything runs.

The same imports can be called directly from a Python code cell in this JupyterLab notebook. A direct call returns a scheduling receipt, not the successor output.

Call `add_code_cell_and_execute` with `content="print(sum(scores) / len(scores))"`. Briefly say what will run. The new ordinary code cell should execute after this answer finishes; do not start another AI question.

## Run an existing identified cell, then ask again

The next code cell has ID `handoff-existing-code`. You may execute it manually first; the handoff still creates a distinct later execution with its own output. Leave its source unchanged after the AI schedules it.

```python
print(f"mean={sum(scores) / len(scores):.1f}")
```

Call `run_and_prompt` with `cell_id="handoff-existing-code"` and `prompt="What mean did the immediately preceding identified code execution print? Explain using its actual run result."` Do not infer the value from an old cell output.

## Let a new AI question choose code

`prompt_and_run` creates a separate tagged question. That new question must explicitly call a handoff tool with one code target. If it does not, no code is scheduled.

Call `prompt_and_run` with `prompt="Create and run one short code cell that prints the maximum of the live scores list. Use add_code_cell_and_execute with the new source; do not only suggest code in prose."`

## Try a direct Python request

Run the next code cell only in JupyterLab with this experimental extension. It requests a new code cell and returns immediately; check the ordinary code cell that appears and runs after this one finishes. In a headless kernel without the frontend, the helper cannot schedule notebook work.

```python
receipt = add_code_cell_and_execute(content="print('direct helper ran after its caller')")
print(receipt)
```

A scheduled cell uses the current notebook and kernel. If the target is edited, deleted, or the session changes before dispatch, the chain stops. Code errors also stop a following question. Keep answer prevents replay of completed AI handoffs, and a new manual run is an intentional new request.

# Inspect a disposable saved notebook

The setup writes a small saved notebook to a temporary folder. The ID `catalog-note` is a real saved cell ID. These tools read the saved file, so unsaved edits to an open notebook are outside their scope.


```python
from pathlib import Path
from tempfile import TemporaryDirectory
import json
from nbinlineai.tools import list_notebooks, find_notebook_cells, read_notebook_cell, search_notebooks, notebook_outline

catalog_temp = TemporaryDirectory(prefix="nbinlineai-saved-")
catalog_root = Path(catalog_temp.name)
catalog_book = catalog_root / "lesson.ipynb"
catalog_book.write_text(json.dumps({"cells": [{"cell_type": "markdown", "id": "catalog-note", "metadata": {}, "source": ["# Observation\n", "Three visits."]}], "metadata": {}, "nbformat": 4, "nbformat_minor": 5}), encoding="utf-8")
print("Saved notebook:", catalog_book)
```

## list_notebooks


```python
print(list_notebooks(str(catalog_root)))
```

## find_notebook_cells


```python
print(find_notebook_cells(str(catalog_book), "visits"))
```

## read_notebook_cell


```python
print(read_notebook_cell(str(catalog_book), "catalog-note"))
```

## search_notebooks


```python
print(search_notebooks("visits", path=str(catalog_root)))
```

## notebook_outline


```python
print(notebook_outline(str(catalog_book)))
```

**Unreleased source example:** these tools are not included in PyPI 0.1.10 yet.

This lesson uses fastcore-backed documentation and file helpers. Run the setup cells first. The files live in a disposable temporary directory, separate from your notebook. Relative tool paths normally use the kernel's working directory; this example supplies an explicit directory.

Direct Python calls need no API key. The AI questions use your configured provider. Keep answer prevents a completed question from repeating its file edits; cancelling a question does not undo edits already made.

```python
from pathlib import Path
from tempfile import TemporaryDirectory
from nbinlineai.fastcore_tools import (
    show_doc, path_info, list_files, view_file, create_file,
    file_str_replace, file_insert_line, file_replace_lines,
)
from nbinlineai.tools import tools_markdown

study_workspace = TemporaryDirectory(prefix="nbinlineai-lesson-")
study_folder = study_workspace.name
study_path = Path(study_folder) / "study-note.txt"
create_file(str(study_path), "Status: planned\nTopic: pollinator visits\n")
print("A disposable study-note.txt is ready.")

def lesson_rate(
    visits: int,  # Number of observed visits.
    flowers: int = 10,  # Number of flowers observed.
) -> float:  # Visits per flower.
    """Calculate the visit rate for a small field observation."""
    return visits / flowers
```

## Formatted documentation

Run the next cell to see a signature with the parameter and return descriptions from fastcore-style comments. The documented function is not called. The same result is returned as Markdown text when an AI calls the tool.

```python
show_doc("lesson_rate")
```

```python
def compare_visits(first: int, second: int) -> int:
    """Compare two observations.

    Parameters
    ----------
    first : int
        Visits in the first observation.
    second : int
        Visits in the second observation.

    Returns
    -------
    int
        The difference between the observations.
    """
    return first - second

show_doc("compare_visits")
```

```python
show_doc("dumps", module="json")
```
